### Creating string

```python
a='Ayush'
print(a)

b="It's rainning today...." # we will use double quotes to avoid syntax error
print(b)

#for multilline string we can use triple quotes
c='''This is a multiline string.
WHERE WE CAN WRITE MULTIPLE LINES OF TEXT WITHOUT ANY ISSUES.'''
print(c)

# We can aslo usse triple double quotes for multiline sting
d="""This is also a multiline string.
WHERE WE CAN WRITE MULTIPLE LINES OF TEXT WITHOUT ANY ISSUES."""
print(d)
```

### Accessing substring from a string

```python
#indexing
s='Ayush'
#python will assign index to each character in the string starting from 0
print(s[0]) # will print A
print(s[1]) # will print y
print(s[2]) # will print u
print(s[3]) # will print s 
print(s[4]) # will print h
```

```python
#negative indexing
#python will assign negative index to each character in the string starting from -1
s='Ayush'
print(s[-1]) # will print h
print(s[-2]) # will print s
print(s[-3]) # will print u 
print(s[-4]) # will print y
print(s[-5]) # will print A
```

```python
#slicing

#print(s[start:end:step])
s='Ayush'
print(s[0:3]) # will print Ayu
print(s[1:4]) # will print yus

print(s[:3]) # will print Ayu
print(s[2:]) # will print ush

print(s[-4:-1]) # will print yus
print(s[1:4:1]) # will print yus
print(s[0:5:2]) # will print Auh
```

### Editing and deleting in string

### 2.Relational operations

```python
#Relational operations on strings

print("ayush"=="Ayush") # this will return False because the strings are case sensitive
print("ayush"!="Ayush") # this will return True because the strings are case sensitive
print("ayush"<"Ayush") # this will return False because the strings are case sensitive
print("ayush">"Ayush") # this will return True because the strings are case sensitive
```

### min() and max()

```python
#min() will returns the minimum character in the string based on ASCII value
print(min('Ayushkumar'))
#max() will returns the maximum character in the string based on ASCII value
print(max('Ayushkumar'))
```

### sorted()


```python
print(sorted('Ayushkumar')) #this will return a list of characters in the string sorted in ascending order based on ASCII value
```

### Capitalize()/Title()/Upper()/lower()/swapcase()


```python
#capitalize()
print('ayush kumar singh'.capitalize())
# it will return the string with the first character capitalized and the rest of the characters in lowercase

#Title ()
print('ayush kumar singh'.title())
#it will return the string with the first character of each word capitalized and the rest of the characters in lowercase

#Upper()
print('ayush kumar singh'.upper())
#it will return the string with all characters in uppercase

#lower()
print('AYUSH KUMAR SINGH'.lower())
#it will return the string with all characters in lowercase

#swapcase()
print('Ayush Kumar Singh'.swapcase())
#it will return the string with all uppercase characters converted to lowercase and all lowercase characters converted to
```

### count()/find()/index()


```python
#count() : it will return the number of occurrences of a substring in the string
print('ayush kumar singh'.count('a'))
```

```python
#find() : it will return the index of the first occurrence of a substring in the string
print('ayush kumar singh'.find('kumar')) # it will return 4
#find() will return -1 if the substring is not found in the string
print('ayush kumar singh'.find('z')) # it will return -1
```

```python
#index() : it will return the index of the first occurrence of a substring in the string
print('ayush kumar singh'.index('h')) # it will return 1
#index() will raise an error if the substring is not found in the string
print('ayush kumar singh'.index('z')) # it will raise an error because 'z' is not found in the string
```

### endswith()/startwith()


```python
#endswith() : it will return True if the string ends with the specified substring, otherwise it will return False
print('ayush kumar singh'.endswith('singh')) # it will return True

#startswith() : it will return True if the string starts with the specified substring, otherwise it will return False
print('ayush kumar singh'.startswith('ayush')) # it will return True
```

### split()/join()


```python
#split() : it will split the string into a list of substrings based on the specified delimiter
s='ayush kumar singh'
print(s.split()) # it will split the string into a list of substrings based on the space character 
print(s.split('u')) # it will split the string into a list of substrings based on the substring 'u'
```

```python
#join() : it will join the elements of a list into a single string with the specified delimiter
a=['ayush', 'kumar', 'singh']
print(' '.join(a))
print('-'.join(['ayush', 'kumar', 'singh']))
```

### Replace()

```python
s='hi my name is ayush'
s.replace('ayush', 'ayush kumar singh') # it will return a new string with all occurrences of the substring 'ayush' replaced with the substring 'kumar'
```

### strip()


```python
#strip() : it will remove the leading and trailing whitespace characters from the string
s='   ayush kumar singh   '
s.strip() # it will return a new string with the leading and trailing whitespace characters removed 
```
