## Literals
### Litreals  are fixed values written directly in a Python program.

```python
a=0b1010 #binary literal
print(a)

b=100 #decimal literal

c=0o710 #octal literal
print(c)

d=0x12c #Hexadecimal literal
print(d)

e=10.5e3 # float literal  1.5*10^3
print(e)

f=1.3j #complex litral 
print(f)
```

## None litral
### The None literal in Python represents the absence of a value or a null value.
