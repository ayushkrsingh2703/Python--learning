## elif
elif stands for "Else If".

In Python, the elif statement is used when we have to check multiple conditions in a program.


```python
#elif grade example
marks=int(input("enter your marks:"))
if marks>=90:
    print("Grade A")
elif marks>=80 and marks <90:
    print("Grade B")
elif marks>=70 and marks<80:
    print("Grade C")
else:
    print("Fail")
```
