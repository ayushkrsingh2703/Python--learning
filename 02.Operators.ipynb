## Operators

```python
# Arithmetic operators
print(3+4) #Addition

print(3-4)#Subtraction

print(3*4) #Multiplication

print(3/4) #Division

print(3//4) #Floor Division

print(3%4) #Modulus

print(3**4) #Exponentiation
```

```python
#Relational operators:Relational operators are used to compare two values.


print(3>4) #Greater than
print(3<4) #Less than
print(3>=4) #Greater than or equal to
print(3<=4) #Less than or equal to
print(3==4) #Equal to
print(3!=4) #Not equal to

```

```python
#Logical operators:Logical operators are used to combine conditional statements.    
print(3>4 and 4>3) #Logical AND
print(3>4 or 4>3) #Logical OR
print(not(3>4)) #Logical NOT
```

```python
#Bitwise operators : Bitwise operators are used to perform bit-level operations on integers.
print(3&2) #Bitwise AND
print(3|2) #Bitwise OR
print(3^2) #Bitwise XOR
print(~3)  #Bitwise NOT
print(3<<2) #Left shift
print(3>>2) #Right shift
```

```python
#Assignment operators:Assignment operators are used to assign values to variables.
a=3 #assigment operator
print(a)
a+=3 #Add AND
print(a)
a-=3 #Subtract AND
print(a)
a*=3 #Multiply AND
print(a)
a/=3 #Divide AND
print(a)
a%=3 #Modulus AND
print(a)
a//=3 #Floor Division AND
print(a)
```

```python
# Membership operators:Membership operators are used to test whether a value or variable is found in a sequence (string, list, tuple, set and dictionary).
x = "Hello World"
print("Hello" in x)  # True
print("World" not in x)  # False
```

```python
# identity operators:Identity operators are used to compare the objects, not if they are equal, but if they are actually the same object, with the same memory location:
x1 = ["apple", "banana"]    
x2 = ["apple", "banana"]    
x3 = x1

print(x1 is x2)  # False
print(x1 is x3)  # True
print(x1 == x2)  # True
```
