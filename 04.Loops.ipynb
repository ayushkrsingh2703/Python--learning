## Loops in python
A loop in Python is used to execute a block of code repeatedly until a given condition is satisfied.
Two types of loops in python :
1.for loop 
2.While loop

## 1.While loop



```python
#program to print table 
num=int(input("Enter a number:"))

i=1
while i<11:
    print(num, "X",i,"=",num*i )
    i+=1
```

### loops with else
In Python, we can use an else block with loops (for and while).

The else block executes when the loop completes normally without encountering a break statement.

```python
#loop with esle
i=5
while i<11:
    print(i)
    i+=1
else:
    print("loop is over")
```

### Example program :
The current population of a town is 10000.The population of the town is increasing at the rate of 10% per year .You have to write a program to find out the population at the end of each of the last 10 year.

```python
population=10000

for i in range(1,11):
    population=population+population*10/100
    print("Population at the end of Year:",i,"is :",round(population))
```

```python
#Sequence sum
#1/1! + 2/2! +3/3!...............

n=int(input("Enter value of n:"))

result=0
fact=1

for i in range(1,n+1):
    fact=fact*i
    result=result+i/fact
    
print(result)

```

### Pattern printing 

```python
# Q1
# *
# * *
# * * *
# * * * *
# * * * * *


rows=int(input("Enter no of rows:"))

for i in range(1,rows+1):
    for j in range(1,i+1):
        print("*",end="")
    print()
```

```python
# Q2
# * * * * *
# * * * *
# * * *
# * *
# *

rows=int(input("Enter the no. of ROWs:"))
for i in range(rows,0,-1):
    for  j in range (i):
        print('*',end='')

    print()
```

```python
# Inverted Number Triangle
row=int(input('enter the no of rows:'))

for i in range (row,0,-1):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
n=int(input("enter the no of rows:"))
for i in range (1,n+1):
    for j in range (1,i+1):
        print(j,end="")
    for k in range (i-1,0,-1):
        print(k,end="")
    print()
```

```python
#pyramid pattern
n=5
for i in range (1,n+1):
    #first we will print the spaces
    for j in range (n-i,0,-1):
        print(" ",end="")

    #then here we will print the stars
    for k in range(2*i-1):
        print("*",end="")
    
    print()
```

## Loop control statements:
Loop control statements are used to change the normal execution of a loop. They help us control how and when a loop executes.


Python has three main loop control statements:

1. break

2. continue

3. pass



### break


```python
for i in range (1,11):
    if i==7:
        break # it breakes the loops when condition is satisfied
    print(i)
```

### continue

```python
for i  in range (1,11):
    if i==5:
        continue #it skip the current iteratioin if the condition is satisfied
    
    print(i)

    
```

### pass 


```python
for i in range (1,11):
    if i==5:
        pass # it does nothing ,it only work as a placeholder and we can fill it in the future .
    print(i)
```
